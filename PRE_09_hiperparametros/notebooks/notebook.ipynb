```python
# APROXIMACION INGENUA
# ----------------------------------------------------------------
# Empaquetado del entrenamiento del modelo


def train_estimator(alpha=0.5, l1_ratio=0.5, verbose=1):

    import os
    import pickle

    import pandas as pd
    from sklearn.linear_model import ElasticNet
    from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
    from sklearn.model_selection import train_test_split

    INPUT_FILE = "../data//winequality-red.csv"
    OUTPUT_FOLDER = "../submission"
    ESTIMATOR_FILE = f"{OUTPUT_FOLDER}/estimator.pkl"

    df = pd.read_csv(INPUT_FILE)

    y = df["quality"]
    x = df.copy()
    x.pop("quality")

    (x_train, x_test, y_train, y_test) = train_test_split(
        x,
        y,
        test_size=0.25,
        random_state=0,
    )

    estimator = ElasticNet(alpha=alpha, l1_ratio=l1_ratio, random_state=12345)

    estimator.fit(x_train, y_train)
    y_pred = estimator.predict(x_test)

    mse = mean_squared_error(y_test, y_pred)
    mae = mean_absolute_error(y_test, y_pred)
    r2 = r2_score(y_test, y_pred)

    if verbose > 0:
        print(estimator, ":", sep="")
        print(f"  MSE: {mse}")
        print(f"  MAE: {mae}")
        print(f"  R2: {r2}")

    if not os.path.exists(OUTPUT_FOLDER):
        os.makedirs(OUTPUT_FOLDER)

    if not os.path.exists(ESTIMATOR_FILE):
        saved_estimator = None
    else:
        with open(ESTIMATOR_FILE, "rb") as file:
            saved_estimator = pickle.load(file)

    if saved_estimator is None or estimator.score(
        x_test, y_test
    ) > saved_estimator.score(x_test, y_test):
        with open(ESTIMATOR_FILE, "wb") as file:
            pickle.dump(estimator, file)
```

```python


def load_data():

    import pandas as pd

    INPUT_FILE = "../data/winequality-red.csv"

    df = pd.read_csv(INPUT_FILE)

    y = df["quality"]
    x = df.copy()
    x.pop("quality")

    return x, y
  
```

```python
# Particionamiento de datos


def make_train_test_split(x, y):

    from sklearn.model_selection import train_test_split

    (x_train, x_test, y_train, y_test) = train_test_split(
        x,
        y,
        test_size=0.25,
        random_state=0,
    )
    return x_train, x_test, y_train, y_test
  
  
  

  
```

```python
# Cálculo de metricas de evaluación


def eval_metrics(y_true, y_pred):

    from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

    mse = mean_squared_error(y_true, y_pred)
    mae = mean_absolute_error(y_true, y_pred)
    r2 = r2_score(y_true, y_pred)

    return mse, mae, r2  
```

```python
# Reporte de métricas de evaluación


def report(estimator, mse, mae, r2):

    print(estimator, ":", sep="")
    print(f"  MSE: {mse}")
    print(f"  MAE: {mae}")
    print(f"  R2: {r2}")  
  
```

```python
# Almacenamiento del modelo


def save_best_estimator(estimator):

    import os
    import pickle

    OUTPUT_FOLDER = "../submission"
    ESTIMATOR_FILE = f"{OUTPUT_FOLDER}/estimator.pkl"

    if not os.path.exists(OUTPUT_FOLDER):
        os.makedirs(OUTPUT_FOLDER)

    with open(ESTIMATOR_FILE, "wb") as file:
        pickle.dump(estimator, file)  
```

```python
# Carga del modelo


def load_best_estimator():

    import os
    import pickle

    OUTPUT_FOLDER = "../submission"
    ESTIMATOR_FILE = f"{OUTPUT_FOLDER}/estimator.pkl"

    if not os.path.exists(ESTIMATOR_FILE):
        return None
    with open(ESTIMATOR_FILE, "rb") as file:
        estimator = pickle.load(file)

    return estimator
  
```

```python
 # Entrenamiento


def train_estimator(alpha=0.5, l1_ratio=0.5, verbose=1):

    from sklearn.linear_model import ElasticNet

    x, y = load_data()
    x_train, x_test, y_train, y_test = make_train_test_split(x, y)
    estimator = ElasticNet(alpha=alpha, l1_ratio=l1_ratio, random_state=12345)
    estimator.fit(x_train, y_train)
    mse, mae, r2 = eval_metrics(y_test, y_pred=estimator.predict(x_test))
    if verbose > 0:
        report(estimator, mse, mae, r2)

    best_estimator = load_best_estimator()
    if best_estimator is None or estimator.score(x_test, y_test) > best_estimator.score(
        x_test, y_test
    ):
        best_estimator = estimator
       

    save_best_estimator(best_estimator)
```

```python
import numpy as np
from sklearn.linear_model import ElasticNet
from sklearn.datasets import make_regression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# Lista para guardar resultados
results = []

# Función que entrena y evalúa el modelo
def train_estimator(alpha, l1_ratio, verbose=0):
    # Datos de ejemplo
    X, y = make_regression(n_samples=200, n_features=20, noise=0.1, random_state=12345)

    # Modelo ElasticNet
    model = ElasticNet(alpha=alpha, l1_ratio=l1_ratio, random_state=12345)
    model.fit(X, y)

    # Evaluación
    y_pred = model.predict(X)
    mse = mean_squared_error(y, y_pred)
    mae = mean_absolute_error(y, y_pred)
    r2 = r2_score(y, y_pred)

    if verbose:
        print(f"alpha={alpha:.4f}, l1_ratio={l1_ratio:.4f}, "
              f"MSE={mse:.4f}, MAE={mae:.4f}, R2={r2:.4f}")

    return mse, mae, r2

# Búsqueda iterativa de hiperparámetros
def make_hyperparameters_search(alphas, l1_ratios):
    for alpha in alphas:
        for l1_ratio in l1_ratios:
            mse, mae, r2 = train_estimator(alpha=alpha, l1_ratio=l1_ratio, verbose=0)
            results.append((alpha, l1_ratio, mse, mae, r2))

# Función para mostrar el mejor ajuste
def check_estimator():
    # Seleccionamos el mejor por MSE
    best = min(results, key=lambda x: x[2])
    print("Mejor combinación encontrada:")
    print(f"alpha={best[0]:.4f}, l1_ratio={best[1]:.4f}")
    print(f"MSE={best[2]:.4f}, MAE={best[3]:.4f}, R2={best[4]:.4f}")

# Ejecución
alphas = np.linspace(0.0001, 0.5, 10)
l1_ratios = np.linspace(0.0001, 0.5, 10)

make_hyperparameters_search(alphas, l1_ratios)
check_estimator()
```

```python
# Búsqueda usando GridSearchCV
def train_estimator(alphas, l1_ratios, n_splits=5, verbose=1):
    from sklearn.linear_model import ElasticNet
    from sklearn.model_selection import GridSearchCV, train_test_split
    from sklearn.datasets import make_regression
    from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

    # Datos de ejemplo
    X, y = make_regression(n_samples=500, n_features=20, noise=0.1, random_state=12345)
    x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=12345)

    # Búsqueda de parámetros con validación cruzada
    estimator = GridSearchCV(
        estimator=ElasticNet(random_state=12345),
        param_grid={"alpha": alphas, "l1_ratio": l1_ratios},
        cv=n_splits,
        refit=True,
        verbose=0,
    )

    estimator.fit(x_train, y_train)
    best = estimator.best_estimator_

    # Evaluación
    y_pred = best.predict(x_test)
    mse = mean_squared_error(y_test, y_pred)
    mae = mean_absolute_error(y_test, y_pred)
    r2 = r2_score(y_test, y_pred)

    if verbose > 0:
        print(f"Mejor combinación: alpha={best.alpha:.4f}, l1_ratio={best.l1_ratio:.4f}")
        print(f"MSE={mse:.4f}, MAE={mae:.4f}, R2={r2:.4f}")

    return best
```
