```python
DATA = "../data"
OUTPUT = "../submission"
```

```python
import nltk

nltk.download("punkt_tab")
nltk.download("stopwords")
```

```python
# Carga de datos


def load_data():

    import glob

    sequence = []
    files = sorted(glob.glob(f"{DATA}/*"))
    for file in files:
        with open(file, "rt", encoding="utf-8") as f:
            sequence.append((file, f.read()))

    return sequence


for file, text in load_data():
    print(f"{file}  {text[:70]}")
```

```python
# Limpieza del texto:
# - une las líneas;
# - elimina espacios repetidos;
# - convierte a minúsculas.


def clean_text(sequence):

    import re

    cleaned_sequence = []
    for file, text in sequence:
        text = re.sub(r"\n", " ", text)
        text = re.sub(r"\s+", " ", text)
        text = text.strip().lower()
        cleaned_sequence.append((file, text))

    return cleaned_sequence


for file, text in clean_text(load_data()):
    print(f"{file}  {text[:70]}")
```

```python
# Tokenización


def tokenize(sequence):

    from nltk.tokenize import word_tokenize

    tokenized_sequence = []
    for file, text in sequence:
        tokenized_sequence.append((file, word_tokenize(text)))

    return tokenized_sequence


for file, tokens in tokenize(clean_text(load_data())):
    print(f"{file}  {' '.join(tokens)[:70]}")
```

```python
# Remoción de datos ruidosos (Opción A):
# conserva sólo tokens alfabéticos; puede perder tokens como "pc3" o "co2".


def filter_tokens_a(sequence):

    filtered_sequence = []
    for file, tokens in sequence:
        filtered_tokens = [token for token in tokens if token.isalpha()]
        filtered_sequence.append((file, filtered_tokens))

    return filtered_sequence


for file, tokens in filter_tokens_a(tokenize(clean_text(load_data()))):
    print(f"{file}  {' '.join(tokens)[:70]}")
```

```python
# Remoción de datos ruidosos (Opción B):
# reemplaza los caracteres no alfabéticos y conserva el resto del token.


def filter_tokens_b(sequence):

    import re

    filtered_sequence = []
    for file, tokens in sequence:
        filtered_tokens = [re.sub(r"[^a-zA-Z\s]", " ", token) for token in tokens]
        filtered_tokens = [re.sub(r"\s+", " ", token).strip() for token in filtered_tokens]
        filtered_tokens = [token for token in filtered_tokens if token != ""]
        filtered_sequence.append((file, filtered_tokens))

    return filtered_sequence


for file, tokens in filter_tokens_b(tokenize(clean_text(load_data()))):
    print(f"{file}  {' '.join(tokens)[:70]}")
```

```python
# Remoción de stopwords


def remove_stopwords(sequence):

    from nltk.corpus import stopwords

    stop_words = set(stopwords.words("english"))

    filtered_sequence = []
    for file, tokens in sequence:
        filtered_tokens = [token for token in tokens if token not in stop_words]
        filtered_sequence.append((file, filtered_tokens))

    return filtered_sequence


for file, tokens in remove_stopwords(filter_tokens_b(tokenize(clean_text(load_data())))):
    print(f"{file}  {' '.join(tokens)[:70]}")
```

```python
# Almacenamiento de resultados


def save_data(sequence):

    import os
    import textwrap

    if not os.path.exists(OUTPUT):
        os.makedirs(OUTPUT)

    for file, tokens in sequence:
        filename = os.path.basename(file)
        with open(f"{OUTPUT}/{filename}", "wt", encoding="utf-8") as f:
            f.write(textwrap.fill(" ".join(tokens), width=70))
```

```python
# Ejecución del proceso completo


def run():

    sequence = load_data()
    sequence = clean_text(sequence)
    sequence = tokenize(sequence)
    sequence = filter_tokens_b(sequence)
    sequence = remove_stopwords(sequence)
    save_data(sequence)


run()
```
